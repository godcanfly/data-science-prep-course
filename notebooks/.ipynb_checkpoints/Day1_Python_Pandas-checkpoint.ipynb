## Part 1: Python Crash Course

A **variable** is a labeled box that stores a value. Python figures out the type automatically.

```python
# Variables and data types
name = "Alice"
age = 18
height = 1.65
is_student = True

print(name, age, height, is_student)
print("Types:", type(name), type(age), type(height), type(is_student))
```

### Lists
- Ordered collection of items
- `scores[0]` gets the first item
- `scores[-1]` gets the last item
- `append()` adds an item
- `len()` gives the length

```python
# Lists
scores = [85, 90, 78, 92]
print("First score:", scores[0])
print("Last score:", scores[-1])
scores.append(88)
print("After append:", scores)
print("Length:", len(scores))
```

### Dictionaries
Dictionaries store key-value pairs.

```python
# Dictionaries
student = {
    "name": "Alice",
    "age": 18,
    "major": "Data Science"
}
print(student["name"])
student["gpa"] = 3.8
print(student)
```

### Conditions and Loops

```python
# Conditions
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"
print("Grade:", grade)

# For loop
for s in scores:
    print(s)

# While loop
n = 0
while n < 5:
    print(n)
    n += 1
```

### Functions
Functions are reusable blocks of code.

```python
# Functions
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))

def calculate_average(numbers):
    return sum(numbers) / len(numbers)

print("Average score:", calculate_average(scores))
```

## Part 3: Introduction to Pandas

Pandas is the industry-standard library for working with data tables in Python. It is like Excel or Google Sheets, but with code.

```python
import pandas as pd

# Load the Titanic dataset from a URL
url = "https://raw.githubusercontent.com/datasciencedoc/data/master/titanic.csv"
df = pd.read_csv(url)

print("Shape:", df.shape)
```

### Inspecting a DataFrame
Use these commands to understand your data quickly.

```python
# First 5 rows
df.head()
```

```python
# Last 5 rows
df.tail()
```

```python
# Column names
print(df.columns.tolist())
```

```python
# Data types and missing values
df.info()
```

```python
# Summary statistics for numeric columns
df.describe()
```

## Part 4: Selecting and Filtering

### Selecting Columns

```python
# One column -> Series
df["Name"].head()
```

```python
# Multiple columns -> DataFrame
df[["Name", "Age", "Sex", "Survived"]].head()
```

### Selecting Rows by Position

```python
# First row
df.iloc[0]
```

```python
# First 5 rows
df.iloc[0:5]
```

### Filtering Rows by Condition

```python
# Passengers older than 30
df[df["Age"] > 30].head()
```

```python
# Female passengers only
df[df["Sex"] == "female"].head()
```

```python
# Multiple conditions
rich_females = df[(df["Sex"] == "female") & (df["Fare"] > 50)]
rich_females.head()
```

### Sorting

```python
# Oldest first
df.sort_values("Age", ascending=False).head()
```

```python
# Sort by class then age
df.sort_values(["Pclass", "Age"]).head()
```

## Part 5: Transformations and Aggregation

### Adding New Columns

```python
# Family size = siblings + parents + self
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df[["SibSp", "Parch", "FamilySize"]].head()
```

### Dropping Columns

```python
# Drop Cabin column if it exists (too many missing values)
if "Cabin" in df.columns:
    df = df.drop(columns=["Cabin"])
print("Columns after drop:", df.columns.tolist())
```

### Handling Missing Values

```python
# Count missing values per column
df.isnull().sum()
```

```python
# Fill missing ages with the median age
median_age = df["Age"].median()
print("Median age:", median_age)
df["Age"] = df["Age"].fillna(median_age)

# Verify there are no missing ages now
df["Age"].isnull().sum()
```

### Groupby and Aggregation

```python
# Average age by passenger class
df.groupby("Pclass")["Age"].mean()
```

```python
# Survival rate by sex
df.groupby("Sex")["Survived"].mean()
```

```python
# Multiple aggregations at once
df.groupby("Pclass").agg({
    "Age": "mean",
    "Fare": "mean",
    "Survived": "mean"
})
```

## Part 6: Save the Cleaned Dataset

```python
# Save to CSV
output_path = "cleaned_titanic.csv"
df.to_csv(output_path, index=False)
print(f"Saved cleaned data to {output_path}")
```
